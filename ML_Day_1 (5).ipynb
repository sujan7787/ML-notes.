**Datasets**

Training Dataset
- We will train our models using this dataset.

Validation Dataset
- Hyperparamater tuning of the model. We check if we can improve accuracy of the model by changing the parameters.

Test Dataset
- We will test the models using this dataset. Test dataset should be hidden from the model.

Ex : Exam
- Training Dataset --> Books(We read question answer and train our brain)
- Validation Dataset --> Different Books (new ideas)
- Testing Dataset --> Question paper(Brain Test) 85% model is performing well

**Model Performance**
- Accuracy High Better Performance
- Accuracy Low Bad Performance

**Overfitting**

- Train Data --> Model is trained --> Accuracy High --> 95%
- Test Data --> Model is tested --> Accuracy Low --> 60%

**Underfitting**

- Train --> Accuracy Low --> 55%
- Test --> Accuracy Low --> 50%

**Generalized Model**
Accuracy in both training data and testing data is high.

Overfitting --> Low Bias, High Variance


Training Accuracy High --> Low Bias

Testing Accuracy Low --> High Variance

Underfitting --> High Bias, High Variance

Generalized Model --> Low Bias, Low Variance

Bias is related to training data and variance is related to testing data.

- Regression --> Output or Dependent Feature --> Continuous
- Classification --> Output or Dependent Feature --> Categorical
